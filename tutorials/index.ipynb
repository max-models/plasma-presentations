# Examples

Guides as notebooks, executed when the documentation is built, and a complete talk.

- [Figures in slides](/plasma-presentations/examples/figures-in-slides/): plasma-plots figures, interactive in HTML and a PDF in Beamer.
- [Scripts](/plasma-presentations/example-scripts/): [First talk](/plasma-presentations/example-scripts/first-talk/) starts a talk, adds a figure and renders it to HTML and PDF.
