# Writing slides

A talk is a Quarto `.qmd` file next to the `plasma` extension. This notebook starts one and walks through what goes into it; rendering it needs [Quarto](https://quarto.org/docs/get-started/) and, for the PDF, lualatex.

## Start a talk

`new_talk` writes a starter `talk.qmd` and copies the extension to `_extensions/plasma` (the command line does the same: `plasma-presentations new kickoff --title ...`).

````python
import tempfile
from pathlib import Path

from plasma_presentations import install_extension, new_talk

folder = Path(tempfile.mkdtemp()) / "kickoff"
qmd = new_talk(
    folder,
    "Spokes in 3D",
    author="Max Lindqvist",
    institute="Max Planck Institute for Plasma Physics",
    date="2026-12-04",
)
sorted(
    path.relative_to(folder).as_posix() for path in folder.rglob("*") if path.is_file()
)
````

## The header

The YAML header names both formats, so one source renders to HTML slides (`plasma-revealjs`) and to a Beamer PDF (`plasma-beamer`). Format options go below each name, for example `plasma-revealjs: {footer: "SPOKES3D kickoff"}` or `plasma-beamer: {fontsize: 12pt}`.

````python
print(qmd.read_text().split("---\n")[1])
````

## The slides

The body is plain [Quarto Markdown](https://quarto.org/docs/presentations/):

- `## Title` starts a slide, `# Section` a section;
- maths with `$...$` and `$$...$$` (MathJax in HTML, LaTeX in the PDF);
- columns with `:::: {.columns}` and `::: {.column width="50%"}`;
- lists shown one item at a time with `::: {.incremental}`;
- speaker notes with `::: {.notes}`.

The starter talk shows each of them:

````python
print(qmd.read_text().split("---\n", 2)[2])
````

## Add a slide

A slide is text appended to the file, here from Python:

````python
with qmd.open("a") as talk:
    talk.write(
        "\n## Debye shielding\n\n"
        "A charge in a plasma is screened over the Debye length\n\n"
        "$$\\lambda_D = \\sqrt{\\frac{\\varepsilon_0 k_B T_e}{n_e e^2}}.$$\n"
    )
print(qmd.read_text().rsplit("## Summary", 1)[1])
````

## Render

```bash
plasma-presentations build kickoff/talk.qmd            # talk.html and talk.pdf
plasma-presentations build kickoff/talk.qmd --to html  # only the HTML slides
quarto preview kickoff/talk.qmd                        # reload while writing
```

or `plasma_presentations.render(qmd)` from Python.

## A folder of talks

Quarto looks for `_extensions/plasma` next to a talk and in every folder above it, so a folder of talks needs one copy. `install_extension` (or `plasma-presentations install talks/`) puts it there, and updates it after an upgrade of the package. Without Python, `quarto add max-models/plasma-presentations` installs the same extension from GitHub.

````python
talks = folder.parent / "talks"
install_extension(talks)
````
